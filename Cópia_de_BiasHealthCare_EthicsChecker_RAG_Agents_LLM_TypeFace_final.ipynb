```python
import os
import requests

def download(url, save_path):

    # Cria a pasta se não existir
    os.makedirs(os.path.dirname(save_path), exist_ok=True)

    print("Baixando...")
    response = requests.get(url)

    if response.status_code != 200:
        print("Erro ao baixar:", response.status_code)
        return

    # Salva o conteúdo diretamente em arquivo
    with open(save_path, "wb") as f:
        f.write(response.content)

    print(f"Download concluído. Arquivo salvo em: {save_path}")

url = 'https://drive.usercontent.google.com/u/1/uc?id=1tw9yENWP_KSKfdKC0NLI5TGKfG9aknub&export=download'

pathFile = 'dados/estatuto-da-pessoa-idosa.pdf'
download(url, pathFile)
```


Agente que analisa o impacto de uma solução utilizando IA para cada ética classica e retorna  se o impacto é "alto", "medio" ou "baixo".



```python
import re
from typing import TypedDict, List
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_typesafe import Noul, Score, Choice, TypeSafeClassifier

class AgentState(TypedDict):
    messages: List[HumanMessage]
    classe_cuidado_1: str
    descricao_cuidado_1: str
    classe_cuidado_2: str
    descricao_cuidado_2: str
    classe_diferencas: str
    descricao_diferencas: str
    classe_est_idoso: str
    descricao_est_idoso: str
    classe_princ: str
    descricao_princ: str
    classe_estruturado: str
    descricao_estruturado: str
    classe_estruturado_principlismo: str
    descricao_estruturado_principlismo: str
    classe_estruturado_estat_idoso: str
    descricao_estruturado_estat_idoso: str


llm = ChatOpenAI(model="gpt-4o-mini")

def analisa(state: AgentState) -> AgentState:
    texto = state["messages"][-1].content
    texto_casos = texto

    # Expressão regular para encontrar e separar o Caso 1 e o Caso 2
    match = re.search(r'(Caso 1.*?)(Caso 2.*)', texto_casos)

    if match:
       caso_1 = match.group(1).strip()
       caso_2 = match.group(2).strip()
    else:
       # Caso não siga o padrão exato, fazemos um split simples
       partes = texto_casos.split('Caso 2')
       caso_1 = partes[0].strip()
       caso_2 = ('Caso 2 ' + partes[1]).strip() if len(partes) > 1 else ''

    print("--- Variável Caso 1 ---")
    print(caso_1)
    print("\n--- Variável Caso 2 ---")
    print(caso_2)

    ## Análise Cuidados 1
    prompt= f"""Descreva os servicos de cuidado a saude para o caso 1 a seguir.
    Definição: {caso_1}
    """

    state["classe_cuidado_1"] = "Analisado"
    response = llm.invoke(prompt).content.strip()
    state["descricao_cuidado_1"] = response

    ## Análise Cuidados 2
    prompt= f"""Descreva os servicos de cuidado a saude para os caso 2 a seguir. Faça a análise de forma separada.
    Definição: {caso_2}
    """

    state["classe_cuidado_2"] = "Analisado"
    response = llm.invoke(prompt).content.strip()
    state["descricao_cuidado_2"] = response

    ## Análise Diferenças entre os Cuidados 1 e Cuidados 2
    prompt= f"""Destaque as diferenças entre os tratamentos sugeridos.
    Definição Cuidados Caso 1: {state["descricao_cuidado_1"]}
    Definição Cuidados Caso 2: {state["descricao_cuidado_2"]}
    A partir do resultado da análise informe possíveis vieses de gênero
    Se algum medicamento indicado ressalte que nenhuma medicação deve ser utilizada sem prescrição médica
    """
    state["classe_diferencas"] = "Analisado"
    response = llm.invoke(prompt).content.strip()
    state["descricao_diferencas"] = response

    ## Análise Estatuto do idoso
    prompt= f"""Descreva a partir dos cuidados informados para cada caso até 5 items adicionais recomendados pelo estatuto do idoso
    Definição Cuidados Caso 1: {state["descricao_cuidado_1"]}
    Definição Cuidados Caso 2: {state["descricao_cuidado_2"]}
    Definição Estatuto: {data_est_idoso}
    """

    state["classe_est_idoso"] = "Analisado"
    response = llm.invoke(prompt).content.strip()
    state["descricao_est_idoso"] = response

    ## Análise princípios éticos impactados considerando a bioética - Principialismo
    prompt= f"""Descreva a partir dos cuidados informados para cada caso até 5 items adicionais recomendados para mitigar viés de genero a partir da lente ética do principialismo.
    Definição Cuidados Caso 1: {state["descricao_cuidado_1"]}
    Definição Cuidados Caso 2: {state["descricao_cuidado_2"]}
    Definição Principialismo: {data_princ}
    """

    state["classe_princ"] = "Analisado"
    response = llm.invoke(prompt).content.strip()
    state["descricao_princ"] = response

    return state

def analisa_estruturado(state: AgentState) -> AgentState:
    texto_1 = state["descricao_cuidado_1"]
    texto_2 = state["descricao_cuidado_2"]
    state["classe_estruturado"] = "Analisado"

    load_dotenv(override=True)
    classifier = TypeSafeClassifier()

    response = classifier.invoke(
        input={
            "state": f"Definição dos casos: Caso 1 {texto_1} Caso 2 {texto_2}",
            "questions": {
                "years_of_treatment": Score(
                    instructions="How many years of treatment is expected, as of today?",
                    criteria=["None", "2 years", "4 years", "6 years", "8 years", "10+ years"]
                ),
                "bias_demonstrated": Noul(
                    instructions="Comparing the treatments suggested for case 1 and for case 2 does the case demonstrate potential bias in the treatment recomendation?"
                ),
                "healthcare_progression": Choice(
                    instructions="Que tipo de abordagem de cuidado continuado é apresentada? Avalie a recomendação de forma holística, considerando aspectos clínicos, funcionais, sociais e de autonomia e os casos 1 e caso 2.",
                    criteria={
                        "comprehensive_care": "Abordagem integrada, considerando necessidades clínicas, funcionais, sociais e de autonomia",
                        "clinical_focus": "Predomínio de recomendações clínicas e terapêuticas",
                        "social_support": "Ênfase em apoio familiar, social e emocional",
                        "independence_focus": "Ênfase na autonomia, independência e capacidade funcional",
                        "potential_bias": "Recomendação apresenta associações ou pressupostos potencialmente relacionados ao gênero, raça ou outras características individuais",
                        "unclear": "A abordagem de cuidado continuado não está claramente caracterizada"
                    }
                ),
                "primary_bias_profile": Choice(
                    instructions="Selecione a melhor correspondência para um possível viés no perfil de tratamento do caso 1 e do caso 2. Avalie o tratamento sugerido de forma holística, e não apenas com base em aspectos específicos.",
                    criteria={
                        "Viés de gênero": "Homens associados a abordagens técnicas; mulheres, a apoio emocional e preventivo.",
                        "Características sociais atribuídas ao gênero": "Mulheres associadas a maior apoio familiar; homens, a maior valorização da independência.",
                        "Percepção diferenciada de independência": "Homens associados à autonomia; mulheres, à dependência ou resiliência.",
                        "Viés interseccional gênero-raça": "Mulheres negras associadas a possíveis barreiras adicionais decorrentes de gênero e raça.",
                        "Inferências não sustentadas": "Respostas incorporaram características sociais ou clínicas ausentes nos casos apresentados.",
                        "Reconhecimento versus reprodução do viés": "Agentes receberam vieses, mas reproduziram estereótipos nas recomendações.",
                        "Diferenças nas recomendações": "Casos clinicamente equivalentes receberam diferentes ênfases relacionadas ao gênero.",
                        "Generalização populacional": "Conhecimentos gerais sobre gênero foram aplicados a indivíduos específicos.",
                        "Especificidade clínica excessiva": "Foi sugerida medicação sem informações suficientes sobre tratamento ou exames.",
                        "Simplificação de direitos legais": "Direitos da pessoa idosa foram apresentados amplamente, sem contextualização jurídica suficiente.",
                        "Igualdade versus equidade": "Respostas defenderam tratamento igualitário e reconheceram barreiras específicas de determinados grupos.",
                        "Padronização positiva do cuidado": "Recomendações recorrentes incluíram avaliação médica, fisioterapia e adaptações domiciliares.",
                        "Abordagem multidimensional": "Foram considerados aspectos clínicos, funcionais, psicológicos, sociais e ambientais.",
                        "Reconhecimento da autonomia": "Respostas enfatizaram preferências individuais e participação nas decisões de cuidado.",
                        "Segurança e não maleficência": "Agentes recomendaram monitoramento, acompanhamento profissional e cautela diante de riscos.",
                        "Justiça e acesso equitativo": "Respostas mencionaram igualdade, não discriminação e redução de vieses."
                    }
                )
            }
        }
    )

    # Salva o resultado estruturado retornado pelo classificador
    state["descricao_estruturado"] = str(response.nouls) + " | " + str(response.scores) + " | " + str(response.choices)
    return state

def analisa_estruturado_principlismo(state: AgentState) -> AgentState:
    texto_1 = state["descricao_cuidado_1"]
    texto_2 = state["descricao_cuidado_2"]
    state["classe_estruturado_principlismo"] = "Analisado"

    load_dotenv(override=True)
    classifier = TypeSafeClassifier()

    response = classifier.invoke(
        input={
            "state": f"Definição dos casos: Caso 1 {texto_1} Caso 2 {texto_2}",
            "questions": {
                "primary_principlism_profile": Choice(
                    instructions="Selecione a melhor correspondência para os principios da bioética no perfil de tratamento do caso 1 e do caso 2. Avalie os principios de forma holística, e não apenas com base em aspectos específicos. ",
                    criteria={
                        "Respeito à autonomia": "Este princípio refere-se à capacidade de um indivíduo de autodeterminar-se e tomar decisões por conta própria, sem pressões indevidas, coerção ou outras formas de persuasão. Ele se contrapõe à noção de paternalismo, que ocorre quando as ações de um profissional de saúde se sobrepõem aos desejos do paciente — ou deixam de respeitá-los — por acreditar-se que o profissional está mais apto a decidir o que é melhor para o paciente. Como regra geral, o médico não tem o direito de tomar decisões importantes em nome de pacientes capazes. Mesmo quando o médico age no interesse do paciente, é fundamental respeitar as escolhas e os desejos do próprio paciente.O respeito à autonomia é operacionalizado por meio da exigência do consentimento livre e esclarecido, segundo o qual pessoas com capacidade de autodeterminação devem ser plenamente informadas antes de terem seu consentimento solicitado.",
                        "Beneficencia": "Este princípio descreve a obrigação de agir em benefício de outras pessoas. Agir dessa forma pode envolver prevenir ou eliminar danos, ou promover ativamente algum bem (por exemplo, a saúde). O objetivo da ação beneficente é obter o 'melhor' resultado possível dentre várias alternativas. Isso pode envolver uma análise de custo-benefício, na qual a 'melhor' opção é aquela em que os benefícios gerados superam, ao máximo, os custos ou riscos. Em suma, trata-se de agir sempre no melhor interesse do paciente.",
                        "Não maleficência": "Os deveres de não maleficência exigem que nos abstenhamos de causar danos deliberados e que evitemos intencionalmente ações que, previsivelmente, causariam danos. Geralmente, as obrigações de não maleficência são mais rigorosas do que as de beneficência; contudo, também aqui pode ser necessária uma análise de custo-benefício para identificar a melhor ação possível. Em algumas situações, o dano pode ser inevitável; nesses casos, devemos assegurar que os benefícios superem os danos.",
                        "Justiça": "O princípio da justiça exige que façamos o possível para garantir uma distribuição equitativa de custos e benefícios. É possível obedecer aos princípios da não maleficência e da beneficência e, ainda assim, não agir de maneira ética, pois esses dois princípios nada dizem sobre como os benefícios devem ser distribuídos. Em um determinado caso, pode ser que só consigamos proporcionar um grande benefício a algumas pessoas prejudicando levemente os interesses de outras. O princípio da beneficência pode indicar que devemos prosseguir, mas, nesse caso, os benefícios e os custos seriam distribuídos de forma injusta."
                    }
                )
            }
        }
    )

    # Salva o resultado estruturado retornado pelo classificador
    state["descricao_estruturado_principlismo"] = str(response.choices)
    return state



def analisa_estruturado_estat_idoso(state: AgentState) -> AgentState:
    texto_1 = state["descricao_cuidado_1"]
    texto_2 = state["descricao_cuidado_2"]
    state["classe_estruturado_estat_idoso"] = "Analisado"

    load_dotenv(override=True)
    classifier = TypeSafeClassifier()

    response = classifier.invoke(
        input={
            "state": f"Definição dos casos: Caso 1 {texto_1} Caso 2 {texto_2}",
            "questions": {
                "primary_estat_idoso_profile": Choice(
                    instructions="Selecione a melhor correspondência para o estatuto do idoso no perfil de tratamento do caso 1 e do caso 2. Avalie os direitos no estatuto do idoso de forma holística, e não apenas com base em aspectos específicos. ",
                    criteria={
                        "Direito à atenção integral e contínua à saúde": "Art. 15, caput. A pessoa idosa tem direito ao atendimento integral pelo Sistema Único de Saúde (SUS), com acesso universal e igualitário às ações e aos serviços de prevenção, promoção, proteção e recuperação da saúde. Na prática: o cuidado deve ser articulado e contínuo, contemplando as necessidades de saúde da pessoa idosa, inclusive no acompanhamento de doenças crônicas.",
                        "Direito a medicamentos de uso continuado e recursos terapêuticos": "Art. 15, 2o. O Poder Público deve fornecer gratuitamente medicamentos à pessoa idosa, especialmente os de uso continuado, além de próteses, órteses e outros recursos necessários ao tratamento, à habilitação ou à reabilitação. Na prática: esse dispositivo fundamenta o direito de solicitar os medicamentos necessários ao tratamento prolongado pelos serviços públicos de saúde, observadas as regras aplicáveis ao fornecimento.",
                        "Direito a atendimento domiciliar quando necessário": "Art. 15, 1o, inciso IV, e 5º e 6º. A legislação prevê atendimento domiciliar, inclusive internação, para a pessoa idosa que necessite desse cuidado e esteja impossibilitada de se locomover. Também assegura atendimento domiciliar para determinadas perícias e emissão de laudos de saúde quando a pessoa estiver enferma. Na prática: quando a condição clínica e os requisitos legais forem atendidos, a impossibilidade de deslocamento deve ser considerada na organização do cuidado, evitando exigir comparecimentos incompatíveis com a condição da pessoa.",
                        "Direito de participar das decisões sobre o tratamento": "Art. 17. À pessoa idosa que esteja no domínio de suas faculdades mentais é assegurado o direito de optar pelo tratamento de saúde que considerar mais favorável, conforme as alternativas apresentadas. Na prática: a idade, por si só, não retira a autonomia da pessoa. Suas preferências e decisões devem ser respeitadas. Quando ela não puder decidir, o artigo prevê regras específicas para a tomada de decisão.",
                        "Direito a acompanhante durante a internação ou observação": "Art. 16. A pessoa idosa internada ou em observação tem direito a acompanhante, devendo o serviço de saúde proporcionar condições adequadas para sua permanência em tempo integral, segundo o critério médico. Na prática: o acompanhamento pode contribuir para a comunicação com a equipe, a compreensão das orientações e a continuidade dos cuidados após a alta. Se o acompanhamento for considerado impossível, o profissional responsável deve justificar a decisão por escrito."
                    }
                )
            }
        }
    )

    # Salva o resultado estruturado retornado pelo classificador
    state["descricao_estruturado_estat_idoso"] = str(response.choices)
    return state
```

```python
import re
from pprint import pprint

def exibe_resultado_estilo_playground(texto_bruto):
    """
    Formata e exibe o resultado estruturado em um formato de dicionário limpo
    e perfeitamente identado utilizando o pprint (Pretty Printer), sem usar json.dumps.
    """
    #texto_bruto = res.get('descricao_estruturado', '')
    if not texto_bruto:
        print("{}")
        return

    estrutura = {}

    # 1. Parsing do bias_demonstrated (Noul)
    bias_match = re.search(r"'bias_demonstrated':\s*NoulAnswer\(type='noul',\s*noul=([0-9.]+)\)", texto_bruto)
    if bias_match:
        estrutura["bias_demonstrated"] = {
            "type": "noul",
            "noul": float(bias_match.group(1))
        }

    # Helper simples para converter a string de dicionário em dicionário real sem json
    def parse_dict_string(d_str):
        # Substitui padrões de strings simples e converte em dicionário Python de forma segura
        try:
            return eval(d_str.strip())
        except:
            return {}

    # 2. Parsing do years_of_treatment (Score)
    years_match = re.search(r"'years_of_treatment':\s*ScoreAnswer\(type='score',\s*score=([0-9.]+),\s*legend=({.*?}),\s*probabilities=({.*?}),\s*confidence=([0-9.]+)\)", texto_bruto)
    if years_match:
        estrutura["years_of_treatment"] = {
            "type": "score",
            "score": float(years_match.group(1)),
            "legend": parse_dict_string(years_match.group(2)),
            "probabilities": parse_dict_string(years_match.group(3)),
            "confidence": float(years_match.group(4))
        }

    # 3. Parsing do healthcare_progression (Choice)
    prog_match = re.search(r"'healthcare_progression':\s*ChoiceAnswer\(type='choice',\s*choice='(.*?)',\s*probabilities=({.*?}),\s*confidence=([0-9.]+)\)", texto_bruto)
    if prog_match:
        estrutura["healthcare_progression"] = {
            "type": "choice",
            "choice": prog_match.group(1),
            "probabilities": parse_dict_string(prog_match.group(2)),
            "confidence": float(prog_match.group(3))
        }

    # 4. Parsing do primary_bias_profile (Choice)
    profile_match = re.search(r"'primary_bias_profile':\s*ChoiceAnswer\(type='choice',\s*choice='(.*?)',\s*probabilities=({.*?}),\s*confidence=([0-9.]+)\)", texto_bruto)
    if profile_match:
        estrutura["primary_bias_profile"] = {
            "type": "choice",
            "choice": profile_match.group(1),
            "probabilities": parse_dict_string(profile_match.group(2)),
            "confidence": float(profile_match.group(3))
        }

   # 4. Parsing do primary_principlism_profile (Choice)
    profile_match = re.search(r"'primary_principlism_profile':\s*ChoiceAnswer\(type='choice',\s*choice='(.*?)',\s*probabilities=({.*?}),\s*confidence=([0-9.]+)\)", texto_bruto)
    if profile_match:
        estrutura["primary_principlism_profile"] = {
            "type": "choice",
            "choice": profile_match.group(1),
            "probabilities": parse_dict_string(profile_match.group(2)),
            "confidence": float(profile_match.group(3))
        }

# 5. Parsing do primary_estat_idoso_profile (Choice)
    profile_match = re.search(r"'primary_estat_idoso_profile':\s*ChoiceAnswer\(type='choice',\s*choice='(.*?)',\s*probabilities=({.*?}),\s*confidence=([0-9.]+)\)", texto_bruto)
    if profile_match:
        estrutura["primary_estat_idoso_profile"] = {
            "type": "choice",
            "choice": profile_match.group(1),
            "probabilities": parse_dict_string(profile_match.group(2)),
            "confidence": float(profile_match.group(3))
        }


    # Exibe a estrutura construída de forma perfeitamente formatada e identada utilizando pprint
    print("\nDICIONÁRIO ESTRUTURADO (IDENTADO VIA PPRINT):\n")
    pprint(estrutura, indent=4, width=120)
```
